```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("./data/kmeans_input.csv")

muestra = df.sample(
    n=100000,
    random_state=42
)

print(muestra.shape)
```

```python
resultados = []

for k in range(2, 9):
    modelo = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    modelo.fit(muestra)

    resultados.append({
        "k": k,
        "inertia": modelo.inertia_
    })

resultados
```

```python
ks = [r["k"] for r in resultados]
inercias = [r["inertia"] for r in resultados]

plt.plot(ks, inercias, marker="o")
plt.xlabel("Número de clusters (K)")
plt.ylabel("Inercia")
plt.title("Método del codo")
plt.show()
```
